# 04. Model Evaluation & Explainability
Evaluating the trained models, viewing metrics.json, scatter plots, and feature importances.


```python
import json, joblib
import pandas as pd
import matplotlib.pyplot as plt

with open("../models/metrics.json") as f:
    metrics = json.load(f)
print("Metrics Summary:\n", json.dumps(metrics, indent=2))
```

```python
with open("../models/feature_importance.json") as f:
    feat_imp = json.load(f)

plt.figure(figsize=(10, 6))
plt.barh(list(feat_imp.keys())[::-1], list(feat_imp.values())[::-1], color="teal")
plt.xlabel("Importance (%)")
plt.title("Logical Feature Importances")
plt.tight_layout()
plt.show()
```
