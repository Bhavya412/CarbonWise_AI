# 01. Exploratory Data Analysis (EDA) — CarbonWise AI
This notebook performs Exploratory Data Analysis on the dataset `data/carbon_emissions.csv`.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/carbon_emissions.csv")
print("Dataset Shape:", df.shape)
```

```python
df.info()
```

```python
print("Missing values per column:\n", df.isnull().sum())
```

```python
df.describe().T
```

```python
print("Target Distribution:\n", df["CarbonEmission"].describe())
plt.figure(figsize=(8, 4))
sns.histplot(df["CarbonEmission"], kde=True, color="teal")
plt.title("Carbon Emission Distribution (kg CO2e)")
plt.xlabel("Carbon Emission (kg CO2e)")
plt.show()
```
